# Structural credit models (Merton equity ↔ credit)

**Start here:** This deep dive expands on `07_advanced_quant/correlation_and_credit_models.ipynb`.

**In this notebook:** We treat firm equity as a call on assets (Merton 1974), calibrate a structural model from equity or CDS inputs, read distance-to-default / PD / spreads, convert to a hazard curve, simulate asset paths, and price a PIK bond with typed Merton MC.

**Measure caveat:** Bound default probabilities and spreads are **risk-neutral** (pricing measure). They overstate physical EDFs whenever the market price of asset risk is positive — do not feed them into rating or capital analytics that need a real-world PD.


## Concept

Under Merton, firm assets $V$ follow geometric Brownian motion. Debt with face $B$ is a defaultable claim; equity is a call:

$$
E = V e^{-qT} N(d_1) - B e^{-rT} N(d_2)
$$

Distance-to-default and terminal PD:

$$
\mathrm{DD} = \frac{\ln(V/B) + (r - q - \sigma_V^2/2)T}{\sigma_V\sqrt{T}}, \qquad
\mathrm{PD} = N(-\mathrm{DD})
$$

Observables are usually equity value and equity vol; KMV / CreditGrades invert those into $(V, \sigma_V)$.


## Imports

```python
import datetime

from finstack_quant.core.currency import Currency
from finstack_quant.core.money import Money
from finstack_quant.core.types import Rate
from finstack_quant.valuations.instruments import (
    BarrierCrossing,
    Bond,
    MertonMcConfig,
    PikMode,
    PikSchedule,
)
from finstack_quant.valuations.models.credit import (
    AssetDynamics,
    BarrierType,
    CreditState,
    DynamicRecoverySpec,
    EndogenousHazardSpec,
    MertonModel,
    ToggleExerciseModel,
)

print("Imports OK")
```

## Asset-side Merton

Specify $(V, \sigma_V, B, r)$ directly when you already have a firm-value view.


```python
m = MertonModel(100.0, 0.25, 80.0, 0.05)
print("V, sigma_V, B, r, q =", m.asset_value, m.asset_vol, m.debt_barrier, m.risk_free_rate, m.payout_rate)
print("DD(1y) =", round(m.distance_to_default(1.0), 6))
print("PD(1y) =", round(m.default_probability(1.0), 6))
print("spread(5y, R=40%) =", round(m.implied_spread(5.0, 0.40), 6))
```

## Equity back-out and KMV calibration

`try_implied_equity` prices equity as a call on assets. `from_equity` inverts observed $(E, \sigma_E)$ into $(V, \sigma_V)$ (KMV fixed-point).


```python
equity, equity_vol = m.try_implied_equity(1.0)
print("implied E, sigma_E =", round(equity, 4), round(equity_vol, 4))

kmv = MertonModel.from_equity(equity, equity_vol, 80.0, 0.05, 0.0, 1.0)
print("KMV V, sigma_V =", round(kmv.asset_value, 4), round(kmv.asset_vol, 4))
print("round-trip |V-100| =", round(abs(kmv.asset_value - 100.0), 6))
```

## CDS and target-PD calibrators

`from_cds_spread` solves for asset vol given a target CDS (bp). `from_target_pd` solves for the debt barrier given a target cumulative PD.


```python
from_cds = MertonModel.from_cds_spread(200.0, 0.40, 80.0, 0.05, 5.0, 100.0, 0.0)
print("from_cds sigma_V =", round(from_cds.asset_vol, 4), "PD(5y) =", round(from_cds.default_probability(5.0), 6))

from_pd = MertonModel.from_target_pd(100.0, 0.25, 0.05, 0.05, 1.0)
print("from_target_pd B =", round(from_pd.debt_barrier, 4), "PD(1y) =", round(from_pd.default_probability(1.0), 6))
```

## CreditGrades vs KMV

CreditGrades uses a different barrier / dynamics construction (Finger et al.). Same equity inputs need not produce the same PD as classic KMV terminal Merton.


```python
e, s_e, debt = 50.0, 0.30, 40.0
cg = MertonModel.credit_grades(e, s_e, debt, 0.05, 0.10, 0.40)
kmv_same = MertonModel.from_equity(e, s_e, debt, 0.05, 0.0, 1.0)
print("CreditGrades PD(1y) =", round(cg.default_probability(1.0), 8))
print("KMV PD(1y)         =", round(kmv_same.default_probability(1.0), 8))
print("CG V / B =", round(cg.asset_value, 4), round(cg.debt_barrier, 4))
```

## Dynamics and first-passage barriers

`new_with_dynamics` selects terminal vs Black–Cox first-passage barriers and GBM / jump-diffusion / CreditGrades dynamics.


```python
terminal = MertonModel.new_with_dynamics(
    100.0, 0.25, 80.0, 0.05, 0.0,
    BarrierType.terminal(),
    AssetDynamics.geometric_brownian(),
)
first_passage = MertonModel.new_with_dynamics(
    100.0, 0.25, 80.0, 0.05, 0.0,
    BarrierType.first_passage(0.0),
    AssetDynamics.geometric_brownian(),
)
print("terminal PD(1y)      =", round(terminal.default_probability(1.0), 6))
print("first-passage PD(1y) =", round(first_passage.default_probability(1.0), 6))
```

## Horizon term structure and sensitivities


```python
base = MertonModel.from_equity(40.0, 0.45, 80.0, 0.05, 0.0, 1.0)
print("horizon  DD       PD        spread(R=40%)")
for t in (0.5, 1.0, 3.0, 5.0):
    dd = base.distance_to_default(t)
    pd = base.default_probability(t)
    sp = base.implied_spread(t, 0.40)
    print(f"{t:5.1f}y  {dd:8.4f}  {pd:8.6f}  {sp:8.6f}")

print("\nequity-vol bump → PD(1y)")
for sig_e in (0.30, 0.45, 0.60):
    m_bump = MertonModel.from_equity(40.0, sig_e, 80.0, 0.05, 0.0, 1.0)
    print(f"  sigma_E={sig_e:.2f}  PD={m_bump.default_probability(1.0):.6f}")
```

## Hazard curve from structural PDs

`to_hazard_curve` bootstraps a piecewise-constant hazard curve from structural survival probabilities.


```python
curve = base.to_hazard_curve(
    "ACME-HZD",
    datetime.date(2024, 1, 15),
    [1.0, 3.0, 5.0],
    0.40,
)
print("curve id =", curve.id)
for t in (1.0, 3.0, 5.0):
    print(f"  t={t:.0f}y  sp={curve.sp(t):.6f}  lambda={curve.hazard_rate(t):.6f}")
```

## Asset-path simulation

`simulate_paths` uses a seeded PCG RNG. CreditGrades dynamics still evolve assets as GBM — the stochastic barrier enters analytic PD formulas only.


```python
paths = base.simulate_paths(2_000, 48, 4.0, seed=42, antithetic=True)
terminals = [paths.get(i, paths.num_steps) for i in range(paths.num_paths)]
mean_terminal = sum(terminals) / len(terminals)
rn_forward = base.asset_value * (2.718281828459045 ** ((base.risk_free_rate - base.payout_rate) * 4.0))
print("num_paths / steps =", paths.num_paths, paths.num_steps)
print("mean terminal V   =", round(mean_terminal, 4))
print("RN forward V      =", round(rn_forward, 4))
```

## PIK building blocks

Recovery, endogenous hazard, and toggle specs compose into Merton MC configs (and JSON `merton_mc_config` overrides).


```python
recovery = DynamicRecoverySpec.constant(0.40)
hazard = EndogenousHazardSpec.power_law(0.02, 4.0, 2.0)
toggle = ToggleExerciseModel.threshold("leverage", 5.0, "above")
state = CreditState(hazard_rate=0.03, leverage=4.5, distance_to_default=1.2)
print("recovery@100 =", recovery.recovery_at_notional(100.0))
print("hazard@8x    =", round(hazard.hazard_at_leverage(8.0), 6))
print("toggle kind  =", toggle.kind)
print("state JSON   =", state.to_json()[:120], "...")
```

## Typed Merton MC on a PIK bond

`Bond.price_merton_mc` runs structural Monte Carlo with a typed `MertonMcConfig`.
Requires GBM dynamics; amortizing / floating bonds are rejected.

Market calibration of barrier/vol to a quote still goes through the registry JSON path
(`instrument_pricing_overrides` + `price_instrument(..., model="merton_mc")`) — see
`02_pricing/instruments/bonds_and_fixed_income.ipynb`.


```python
merton = MertonModel(100.0, 0.30, 70.0, 0.04)
config = (
    MertonMcConfig(merton)
    .num_paths(1_000)
    .seed(7)
    .antithetic(True)
    .pik_schedule(PikSchedule.uniform(PikMode.pik()))
    .dynamic_recovery(recovery)
    .barrier_crossing(BarrierCrossing.discrete())
)
bond = Bond.fixed(
    "PIK-DEMO",
    Money(100.0, Currency("USD")),
    Rate(0.08),
    datetime.date(2024, 1, 15),
    datetime.date(2029, 1, 15),
    "USD-OIS",
)
result = bond.price_merton_mc(config, 0.04, datetime.date(2024, 1, 15))
print("clean % par     =", round(result.clean_price_pct, 4))
print("expected loss   =", round(result.expected_loss, 6))
print("default rate    =", round(result.path_statistics.default_rate, 4))
print("std error       =", round(result.standard_error, 4))
```

## Takeaways

- Equity is a call on firm assets; KMV `from_equity` / `try_implied_equity` round-trip that duality.
- CreditGrades is a separate barrier construction — do not expect identical PDs to classic Merton for the same equity inputs.
- Bound PDs and spreads are risk-neutral; use a physical drift for real-world EDFs.
- `to_hazard_curve` and `simulate_paths` bridge structural analytics into curve and MC workflows.
- Typed `MertonMcConfig` + `Bond.price_merton_mc` prices PIK/toggle bonds; JSON registry calibration remains in the bonds notebook.
